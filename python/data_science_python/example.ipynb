```python
users = pd.read_csv("./users.csv")
users
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))
plt.bar(users["username"], users["balance"],color="green",width=0.5, label="Balance")
plt.xlabel("Username")
plt.ylabel("Balance")
plt.title("User Balance")
plt.legend()
plt.show()
```

```python
plt.pie(users["balance"], labels=users["username"],autopct='%1.1f%%', startangle=90)
plt.title("User Balance Distribution")
plt.show()
```
