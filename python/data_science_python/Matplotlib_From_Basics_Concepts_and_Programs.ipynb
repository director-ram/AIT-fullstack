## 5. Your First Matplotlib Plot

The simplest line plot can be created using:

```python
plt.plot(x, y)
```

- `x` contains X-axis values.
- `y` contains Y-axis values.
- `plt.plot()` creates the line plot.
- `plt.show()` displays it.

### Example


````python
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.plot(x, y)
plt.show()
````

### How does the program work?

The corresponding values are:

| X | Y |
|---:|---:|
| 1 | 10 |
| 2 | 20 |
| 3 | 15 |
| 4 | 30 |
| 5 | 25 |

Matplotlib plots these points:

`(1,10) → (2,20) → (3,15) → (4,30) → (5,25)`

and connects them with a line.

### Important

`plt.show()` tells Matplotlib to display the created figure.


## 6. Understanding X-axis and Y-axis

A plot normally has two axes.

### X-axis

The horizontal axis.

It commonly represents categories, time or an independent variable.

### Y-axis

The vertical axis.

It commonly represents a measured value or dependent variable.

Example:

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
sales = [100, 150, 120, 180, 200]
```

Here:

- `days` → X-axis
- `sales` → Y-axis


````python
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
sales = [100, 150, 120, 180, 200]

plt.plot(days, sales)
plt.show()
````

## 7. Adding a Title

A title tells the viewer what the graph represents.

### Syntax

```python
plt.title("Your Title")
```


````python
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
sales = [100, 150, 120, 180, 200]

plt.plot(days, sales)
plt.title("Weekly Sales")
plt.show()
````

### How does it work?

```python
plt.plot(days, sales)
```
creates the graph.

```python
plt.title("Weekly Sales")
```
adds the title above the graph.

The title should clearly describe the data.


## 8. Adding X-axis and Y-axis Labels

Use:

```python
plt.xlabel("...")
plt.ylabel("...")
```

These labels tell the viewer what each axis represents.


````python
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
sales = [100, 150, 120, 180, 200]

plt.plot(days, sales)

plt.title("Weekly Sales")
plt.xlabel("Days")
plt.ylabel("Sales")

plt.show()
````

## 9. Adding Markers

A **marker** shows the individual data points on a graph.

### Syntax

```python
plt.plot(x, y, marker="o")
```

Some common markers:

| Marker | Meaning |
|---|---|
| `"o"` | Circle |
| `"s"` | Square |
| `"*"` | Star |
| `"^"` | Triangle |
| `"x"` | X |

Markers are especially useful when you want individual observations to be visible.


````python
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.plot(x, y, marker="o")
plt.title("Line Plot with Markers")
plt.show()
````

## 10. Line Style

The `linestyle` parameter controls the appearance of the connecting line.

Common values:

| Value | Appearance |
|---|---|
| `"-"` | Solid |
| `"--"` | Dashed |
| `":"` | Dotted |
| `"-."` | Dash-dot |


````python
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.plot(x, y, linestyle="--", marker="o")
plt.title("Dashed Line")
plt.show()
````

## 11. Line Width and Color

### Line width

```python
linewidth=3
```

controls how thick the line is.

### Color

```python
color="blue"
```

sets the line color.

You can combine them with markers and line styles.


````python
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.plot(
    x, y,
    color="blue",
    linewidth=3,
    linestyle="-",
    marker="o"
)

plt.title("Customized Line Plot")
plt.show()
````

## 12. Grid

A grid helps the viewer read values from the graph.

### Syntax

```python
plt.grid()
```


````python
months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [100, 150, 130, 180, 220]

plt.plot(months, sales, marker="o")
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid()

plt.show()
````

## 13. Legend

A **legend** identifies different plotted datasets.

First give the plot a label:

```python
plt.plot(x, y, label="Sales")
```

Then display the legend:

```python
plt.legend()
```


````python
months = ["Jan", "Feb", "Mar", "Apr"]
sales = [100, 150, 130, 180]

plt.plot(months, sales, marker="o", label="Sales")

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()

plt.show()
````

## 14. Multiple Lines in One Graph

We can plot multiple datasets on the same axes.

This is useful for **comparison**.

For example, compare two products' monthly sales.


````python
months = ["Jan", "Feb", "Mar", "Apr"]

product_a = [100, 150, 180, 200]
product_b = [80, 120, 160, 190]

plt.plot(months, product_a, marker="o", label="Product A")
plt.plot(months, product_b, marker="o", label="Product B")

plt.title("Product Sales Comparison")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()
plt.grid()

plt.show()
````

### How does it work?

The first `plt.plot()` creates Product A's line.

The second `plt.plot()` creates Product B's line.

Because both plots use the same axes, we can compare them directly.

The `label` values are displayed by `plt.legend()`.


# 15. Bar Chart

A **bar chart** is mainly used to compare values across categories.

### Syntax

```python
plt.bar(categories, values)
```

For example, we can compare marks in different subjects.


````python
subjects = ["Python", "Java", "C", "HTML"]
marks = [90, 75, 80, 95]

plt.bar(subjects, marks)

plt.title("Student Marks")
plt.xlabel("Subjects")
plt.ylabel("Marks")

plt.show()
````

### Line Plot vs Bar Chart

**Line plot:** best for showing trends or changes.

**Bar chart:** best for comparing categories.

For example:

- Monthly sales trend → Line plot
- Marks in different subjects → Bar chart


## 16. Horizontal Bar Chart

Use `plt.barh()` when you want horizontal bars.

### Syntax

```python
plt.barh(categories, values)
```

# 17. Scatter Plot

A **scatter plot** displays individual observations as points.

It is useful for studying the **relationship between two numerical variables**.

### Syntax

```python
plt.scatter(x, y)
```

Example question:

> Does studying more hours relate to higher marks?


````python
hours = [1, 2, 3, 4, 5, 6]
marks = [40, 45, 55, 65, 75, 85]

plt.scatter(hours, marks)

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
````

### How does it work?

Each pair forms one point:

`(1,40), (2,45), (3,55), (4,65), (5,75), (6,85)`

The upward pattern suggests a **positive relationship** in this small example.

Important: a scatter plot can show association or correlation patterns, but it does not by itself prove causation.


# 18. Histogram

A **histogram** shows the distribution of numerical data.

It divides the data into intervals called **bins** and shows how frequently values fall into those intervals.

### Syntax

```python
plt.hist(data)
```


````python
marks = [45, 50, 55, 60, 60, 65, 70, 70, 75, 80, 85, 90]

plt.hist(marks, bins=5)

plt.title("Distribution of Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")

plt.show()
````

### What is frequency?

**Frequency** means how many observations occur in a particular range.

### Histogram vs Bar Chart

**Bar chart**
- Compares categories.
- Bars represent separate categories.

**Histogram**
- Shows numerical distribution.
- Values are grouped into intervals or bins.


# 19. Pie Chart

A **pie chart** represents parts of a whole.

It is useful when we want to show proportions or percentages.

### Syntax

```python
plt.pie(values, labels=labels)
```


````python
subjects = ["Python", "Java", "C", "HTML"]
hours = [40, 20, 25, 15]

plt.pie(hours, labels=subjects)

plt.title("Study Time Distribution")

plt.show()
````

## 20. Showing Percentages in a Pie Chart

The `autopct` parameter controls the percentage text.

```python
autopct="%1.1f%%"
```

Meaning:

- `%1.1f` → display a number with one decimal place
- `%%` → display the percent symbol


````python
subjects = ["Python", "Java", "C", "HTML"]
hours = [40, 20, 25, 15]

plt.pie(
    hours,
    labels=subjects,
    autopct="%1.1f%%"
)

plt.title("Study Time Distribution")

plt.show()
````

## 21. Exploding a Pie Slice

The `explode` parameter separates selected slices from the center.

For example:

```python
explode = [0.1, 0, 0, 0]
```

The first slice is separated slightly.


````python
subjects = ["Python", "Java", "C", "HTML"]
hours = [40, 20, 25, 15]

explode = [0.1, 0, 0, 0]

plt.pie(
    hours,
    labels=subjects,
    explode=explode,
    autopct="%1.1f%%"
)

plt.title("Study Time Distribution")

plt.show()
````

# 22. Figure and Axes

Matplotlib has two important concepts:

### Figure

The complete canvas/window containing the visualization.

### Axes

The actual plotting area where data is drawn.

A common object-oriented approach is:

```python
fig, ax = plt.subplots()
```

Then we use:

```python
ax.plot(...)
```

instead of:

```python
plt.plot(...)
```


````python
fig, ax = plt.subplots()

ax.plot([1, 2, 3], [10, 20, 15])

ax.set_title("My Graph")
ax.set_xlabel("X")
ax.set_ylabel("Y")

plt.show()
````

### Why learn `Figure` and `Axes`?

The `plt` interface is convenient for simple plots.

The object-oriented approach becomes especially useful when a figure contains multiple axes or when building larger visualizations.


# 23. Subplots

**Subplots** allow us to put multiple plots inside one Figure.

### Syntax

```python
fig, ax = plt.subplots(rows, columns)
```

For example:

```python
plt.subplots(1, 2)
```

means:

- 1 row
- 2 columns

So we get two plotting areas side by side.


````python
fig, ax = plt.subplots(1, 2, figsize=(10, 4))

ax[0].plot([1, 2, 3], [10, 20, 15])
ax[0].set_title("Line Plot")

ax[1].bar(["A", "B", "C"], [10, 20, 15])
ax[1].set_title("Bar Chart")

plt.tight_layout()
plt.show()
````

### Understanding `ax[0]` and `ax[1]`

Because there are two Axes objects:

```python
ax[0]
```

refers to the first plot.

```python
ax[1]
```

refers to the second plot.

This lets us control each subplot separately.


# 24. Saving a Plot

We can save a visualization instead of only displaying it.

### Syntax

```python
plt.savefig("filename.png")
```

Useful formats include PNG, JPG and PDF, depending on the intended use.


````python
x = [1, 2, 3, 4]
y = [10, 20, 15, 30]

plt.plot(x, y, marker="o")
plt.title("Sales")

plt.savefig("sales.png", dpi=300, bbox_inches="tight")

plt.show()
````

### Important parameters

`dpi=300` controls the resolution of the saved image.

`bbox_inches="tight"` helps reduce unnecessary whitespace around the figure.


# 25. Complete Demonstration Program

Now let's combine the major concepts:

- Data
- Line plot
- Markers
- Line width
- Title
- X-axis label
- Y-axis label
- Grid
- Legend
- Display


````python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [100, 150, 130, 180, 220]

plt.plot(
    months,
    sales,
    marker="o",
    linestyle="-",
    linewidth=2,
    label="Sales"
)

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.grid()
plt.legend()

plt.show()
````
