```python
# from cProfile import label
# from inspect import markcoroutinefunction


months = ["Jan", "Feb", "Mar", "Apr"]
sale_1 = [100, 150, 130, 180]
sale_2 = [110, 120, 160, 170]
plt.plot(months, sale_1, marker="*", ms=12,label="Sale-1", linewidth=2)
plt.plot(months,sale_2,marker="^",mec="yellow",mfc="red",ms=12,color="black",linewidth=3,linestyle="-.",label="sale-2")

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()
plt.grid()

plt.show()
```

```python
import numpy as np

subjects = ["Python", "Java", "C", "HTML"]
marks = [90, 75, 80, 95]
student = [56, 23, 89, 77]

x = np.arange(len(subjects))
width = 0.40

plt.bar(x - width/2, marks, width, label="Student 1")
plt.bar(x + width/2, student, width, label="Student 2")

plt.title("Student Marks Comparison")
plt.xlabel("Subjects")
plt.ylabel("Marks")

plt.xticks(x, subjects)
plt.legend()

plt.show()
```
